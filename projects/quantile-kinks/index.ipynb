# Quantile kinks & ecological breakpoints

**Status: Completed · Illustrative study**

This constructed example explains how a change in an upper response envelope can describe an ecological breakpoint. It is not an empirical research claim.

## Background & question

Can a continuous change in slope identify a breakpoint when other environmental factors depress some observations?

## Project objective

Describe the location and interpretation of an ecological response kink, using a known generating model as a benchmark.

## Analytical framework

Represent a continuous quantile kink as

$$q_\tau(x)=\beta_0+\beta_1x+\beta_2\max(0,x-c).$$

The slopes before and after the breakpoint are $\beta_1$ and $\beta_1+\beta_2$.

### Inputs, method, and outputs

The selected dataset contains 160 independent synthetic observations. Pressure ranges from 0 to 100. The upper boundary is $b(x)=8-0.02x-0.10\max(0,x-50)$. Independent exponential noise with mean 1.2 is subtracted from this boundary.

The computation was performed separately; this notebook contains only the selected presentation.

## Results

![Synthetic observations and exact upper quantile](assets/quantile-envelope.svg)

*The green line is the exact conditional 0.90 quantile, not a fitted curve. The vertical line marks the planted breakpoint.*

| Quantity | Constructed value |
|---|---|
| Breakpoint | 50 |
| Pre-kink slope | −0.02 |
| Post-kink slope | −0.12 |
| 0.90 quantile below boundary | Approximately 0.126 |

Because the noise distribution is constant along the gradient, the conditional quantile retains the boundary's kink.

## Discussion & limitations

The kink is a change in rate, not an abrupt jump. A statistical breakpoint alone does not establish a causal ecological threshold.

The example assumes one kink, independent observations, and a constant noise distribution. No fitted breakpoint or confidence interval is claimed.

## Conclusion

The constructed model has a breakpoint at 50 and a slope change of −0.10. It demonstrates a complete research presentation without including computation notebooks.
