## Testing function $\phi(D_n)$

From the generated $D_n$ to answer if a breakpoint exists, it is equivalent to the question of whether the $\delta$ parameter is 0 or not. The possible hypotheses are:

$$
H_0: \delta = 0, \qquad H_1: \delta \neq 0
$$

Suppose a **testing function** $\phi(\cdot)$ is responsible for the decision. It takes a dataset $D_n$ and outputs:

$$
\phi(D_n)=
\begin{cases}
1, & \text{reject } H_0,\\
0, & \text{accept } H_0.
\end{cases}
$$

The final decision is supported by statistic evidence through an inference procedure $\mathcal{T}$ applied to $D_n$.

The $\mathcal{P}_{\theta}$ and $\mathcal{S}_{\eta}$ are known in simulation, so we know whether $H_0$ or $H_1$ is closer to the truth. Then we watch how the $\phi(\cdot)$ performs across $D_n$.

Under $H_1$ scenario of $\delta\neq0$, the **power** is

$$
\pi_{\theta,\eta}
=P_{\theta,\eta}\{\phi(D_n)=1\}
=E_{\theta,\eta}[\phi(D_n)],
\qquad \delta\neq0.
$$

Under $H_0$ scenario of $\delta=0$, the same rejection probability is the **Type I error**, which should be controlled near the prescribed level $\alpha$:

$$
\alpha_{\theta,\eta}
=P_{\theta,\eta}\{\phi(D_n)=1\}
=E_{\theta,\eta}[\phi(D_n)],
\qquad \delta=0.
$$

Conclusively, a $\phi(\cdot)$ applies over data $D_n$ generated from the specified $\mathcal{P}_{\theta},\mathcal{S}_{\eta}$ processes and its performance is evaluated under this scenario.

$$
\underbrace{\mathcal{P}_{\theta}}_{\text{true relation}} 
\xrightarrow{\underbrace{\mathcal{S}_{\eta}}_{\text{sampling design}}} 
\underbrace{D_n}_{\text{sample data}}
\xrightarrow{\underbrace{\mathcal{T}}_{\text{detector}}} 
\underbrace{\phi(D_n)}_{\text{decision}}.
$$
