# Cooperation in the prisoner’s dilemma

**Status: Ongoing · Illustrative notebook**

## Background & question

When individual incentives conflict with collective benefits, what can sustain cooperation?

## Model

Two players choose Cooperate (C) or Defect (D). Cells give the row player's and column player's payoffs.

| Your choice | Other: C | Other: D |
|---|---|---|
| C | 3, 3 | 0, 5 |
| D | 5, 0 | 1, 1 |

## Initial findings

### One-shot game

Defection is strictly dominant: it gives 5 instead of 3 against cooperation, and 1 instead of 0 against defection. The unique one-shot Nash equilibrium is mutual defection.

### Repeated-game benchmark

With perfect monitoring, no errors, indefinite repetition, and grim-trigger punishment, cooperation gives $3/(1-\delta)$ while a one-time deviation gives $5+\delta/(1-\delta)$.

Cooperation satisfies the incentive constraint when

$$\frac{3}{1-\delta}\geq 5+\frac{\delta}{1-\delta},\qquad \delta\geq 0.5.$$

This is an analytical benchmark, not evidence that cooperation will occur.
